```python
import numpy as np
import matplotlib as pyplot
from matplotlib import pyplot as plt
```

```python
x = np.array([100,200,300,400,500,600,700,800,900,1000])
y = np.array([11,22,33,44,55,66,77,88,99,200])
x,y
```

```python
plt.scatter(x,y, color = "red", marker = "*", s = 100)

plt.title("Scatter Plot")
plt.xlabel("x-label")
plt.ylabel("y-label")

plt.show()
```
