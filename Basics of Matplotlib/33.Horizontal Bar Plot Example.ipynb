```python
import numpy as np
import matplotlib as pyplot
from matplotlib import pyplot as plt
```

```python
dictionary = ({"Sam":1,"Jhon":2,"Alia":3,"Afia":4,"Ali":5})
dictionary
```

```python
dictionary.keys()
```

```python
dictionary.values()
```

```python
name = list(dictionary.keys())
values = list(dictionary.values())
```

```python
plt.barh(name,values, color = "green")

plt.title("Bar Plot")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.grid(True)

plt.show()
```
