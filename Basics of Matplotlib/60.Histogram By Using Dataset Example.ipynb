```python
import numpy as np
import pandas as pd
import matplotlib as pyplot
from matplotlib import pyplot as plt
```

```python
n = pd.read_csv("Demo Data.csv")
```

```python
n.head()
```

```python
plt.subplot(2,1,1)

plt.hist(n["Weight"], color = "Red", bins = 10)
plt.title("Histogram")
plt.ylabel("y-label")

plt.subplot(2,1,2)

plt.hist(n["Salary"], color = "Green", bins = 10)
plt.xlabel("x-label")
plt.ylabel("y-label")

plt.show()
```
